# EDA Cleaning Feature Engineering
Customer User Journey Analytics

```python
import pandas as pd
df = pd.read_csv('../data/user_realistic.csv')
df.head()
```

```python
df.info()
df.isna().sum()
```

```python
df.columns = [c.strip().lower().replace(' ', '_') for c in df.columns]
df['journey_date'] = pd.to_datetime(df['journey_date'], errors='coerce')
df = df.drop_duplicates()
df['conversion_flag'] = (df['funnel_stage'] == 'Purchase').astype(int)
df['retention_gap'] = df['d1_retained'] - df['d7_retained']
df.head()
```

```python
funnel = df.groupby(['funnel_stage_order','funnel_stage'])['customer_id'].nunique().reset_index(name='users')
funnel['conversion_from_previous'] = funnel['users'] / funnel['users'].shift(1)
funnel
```

```python
df.to_csv('../data/user_journey_clean.csv', index=False)
funnel.to_csv('../data/funnel_summary.csv', index=False)
```

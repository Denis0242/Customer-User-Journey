# EDA Cleaning Notebook — Customer User Journey


```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw_user_journey.csv')
df.head()
df.shape
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
df.columns
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_', regex=False)
df['journey_date'] = pd.to_datetime(df['journey_date'], errors='coerce')
for col in ['revenue','d1_retained','d7_retained','funnel_stage_order']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df = df.drop_duplicates()
df['converted_flag'] = (df['funnel_stage'].str.lower() == 'purchase').astype(int)
df['revenue_user_flag'] = (df['revenue'] > 0).astype(int)
df['retention_status'] = np.where(df['d7_retained'] == 1, 'D7 Retained', np.where(df['d1_retained'] == 1, 'D1 Only', 'Not Retained'))
df.to_csv('../data/cleaned_user_journey.csv', index=False)
```
